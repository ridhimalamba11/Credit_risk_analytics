```python
import pandas as pd
df = pd.read_csv('D:/credit_risk/data/processed/cleaned_data.csv')
df.head()
```

```python
df['Late_Payments'] = (
    (df['PAY_0'] > 0).astype(int) +
    (df['PAY_2'] > 0).astype(int) +
    (df['PAY_3'] > 0).astype(int) +
    (df['PAY_4'] > 0).astype(int) +
    (df['PAY_5'] > 0).astype(int) +
    (df['PAY_6'] > 0).astype(int)
)
df['Late_Payments'].describe()
```

```python
#identifying the risk by calculating the utilization ratio
df['Utilization'] = df['BILL_AMT1'] / df['LIMIT_BAL']
df['Utilization'].head()
df['Utilization'].describe()
```

```python
#creating the columns for total payment
df['Total_Payment'] = (
    df['PAY_AMT1'] +
    df['PAY_AMT2'] +
    df['PAY_AMT3'] +
    df['PAY_AMT4'] +
    df['PAY_AMT5'] +
    df['PAY_AMT6']
)
```

```python
df['Total_Bill'] = (
    df['BILL_AMT1'] +
    df['BILL_AMT2'] +
    df['BILL_AMT3'] +
    df['BILL_AMT4'] +
    df['BILL_AMT5'] +
    df['BILL_AMT6']
)
```

```python
df['Payment_Ratio'] = (
    df['Total_Payment'] /
    (df['Total_Bill'] + 1)
)
```

```python
df['Age_Group'] = pd.cut(
    df['AGE'],
    bins=[20,30,40,50,80],
    labels=['20-30','30-40','40-50','50+']
)
```

```python
df[
[
'Late_Payments',
'Utilization',
'Total_Bill',
'Total_Payment',
'Payment_Ratio'
]
].head()
```

```python
df.to_csv(
'D:/credit_risk/data/processed/final_credit_risk_data.csv',
index=False
)
```
